```python
# Alternate ways - Create a zero tensor
size = (3, 4, 6)
tensor3 = torch.empty(size)
tensor3
```

```python
# initialise a random tensor

random_tensor = torch.rand(size)
random_tensor
```

```python
# zero tensor and ones tensor
zero_tensor = torch.zeros(size)
ones_tensor = torch.ones(size)

zero_tensor, ones_tensor
```

```python
# check datatype of the tensor - default is float32 for any tensor
tensor = torch.rand(1, 2)
print(tensor)
tensor.dtype
```

```python
# create a tensor with specific datatype
tensor = torch.tensor([1,2,3,4], dtype=torch.float16)
print(tensor)
tensor.dtype
```

```python
# change the datatype of a tensor
tensor_another = tensor.type(torch.double)
tensor_another.dtype
```

```python
# convert numpy array to a tensor
example = np.array([[random.randint(1, 9) for _ in range(10)], 
                    [random.randint(1,9) for _ in range(10)]])

# from_numpy shares the same memory as the example
tensor_one = torch.from_numpy(example)

# changes done in the example array does not reflect in this tensor
tensor_two = torch.tensor(example)

example, tensor_one, tensor_two
```

```python
# demonstarting the difference between the two methods
example *= 3

example, tensor_one, tensor_two
```

```python
## creating a tensor using another tensor
tensor_three = torch.ones_like(tensor_one)
tensor_three
```

```python
# accessing element from a tensor
print(tensor2)
print(f"Dim: {tensor2.dim()}")
print(f"Size: {tensor2.size()}")

# print array in the first dim
print(tensor2[1])

# print element using specific index
print(tensor2[0][2])
```

## Basic Tensor Operations

```python
tensor_one = torch.tensor([10, 20, 30])
tensor_two = torch.tensor([10, 10, 10])
```

```python
# Element wise arithmetic operations

# Addition
print(tensor_one + tensor_two)

# Subtraction
print(tensor_one - tensor_two)

# Multiplication
print(tensor_one * tensor_two)

# Division
print(tensor_one / tensor_two)
```

```python
# random tensors (again)
random_tensor = torch.randint(1,9, size=(4,6))
random_tensor
```

```python
# reshape tensors 

y = random_tensor.view(24)
z = random_tensor.view(-1, 2)

print(y, z)
```

```python
# create a tensor by specifying a range
tensor_a = torch.arange(24)
tensor_a = tensor_a.reshape(4,6)
tensor_a
```

```python
# concatenate two arrays
concat = torch.cat((random_tensor, tensor_a), dim=0)
concat
```
