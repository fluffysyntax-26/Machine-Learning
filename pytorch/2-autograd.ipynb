## AutoGrad

$$y=3\sigma(x) + 5$$

$$\frac{\partial y}{ \partial x} = 3*\sigma(x)(1-\sigma(x))$$

$$\frac{\partial y}{ \partial x} = 3*\sigma(2)(1-\sigma(2))=3*0.8808*(1-0.8808)=0.3149$$

```python
import torch
```

### requires_grad (The "Record" Button)
This is a boolean flag that tells PyTorch whether or not to track operations on this tensor to build a Computational Graph.

### is_leaf (The "Origin" Point)
The "Leaf" concept comes from tree structures. A leaf tensor is a base node in the computational graph—it wasn't created by an operation involving other tensors that are being tracked.

```python
x = torch.tensor(2.0)
x.requires_grad,x.is_leaf
```

```python
y = 3 * torch.sigmoid(x) + 5
y.requires_grad,y.is_leaf
```

```python
# explicitly set requires_grad to true while creating a tensor
x = torch.tensor(2.0, requires_grad=True)

x.requires_grad, x.is_leaf
```

```python
y = 3 * torch.sigmoid(x) + 5
y
```

```python
y.requires_grad,y.is_leaf
```

```python
print(x.grad_fn) # doesn't have a grad function
print(y.grad_fn) # does contain a pointer 
```

```python
print(x.grad)
y.backward()
print(x.grad) # dy/dx
```

```python
x.grad.zero_() # resets gradients, else they would get accumulated
y = 3 * torch.sigmoid(x) + 5
y.backward()
x.grad
```
