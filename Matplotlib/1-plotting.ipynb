```python
import matplotlib.pyplot as plt
import numpy as np
import random
```

```python
# lay down the graph and use show() to display the image
plt.plot()
plt.show()
```

```python
# plot a graph which takes a simple list as an argument
nums = [x for x in range(1, 6)]
plt.plot(nums)
plt.show()
```

```python
# plot with x and y axis defined
x = [x for x in range(1, 11)]
y = [y**2 for y in range(1, 11)]

plt.title("Numbers and their squares")
plt.xlabel("X")
plt.ylabel("x^2")
plt.axis([0, 10,0,100])
plt.plot(x, y, 'ro-', linewidth=3.0) # markers for better viz
plt.grid(True)
plt.show()
```

```python
# plot multiple lines in a single graph

x = np.array([x for x in range(1, 11)])

plt.plot(x, x**2, 'r-', x, x**3, 'g-', x, x**4, 'b-')
plt.title("Numbers and their exponentiation")
plt.xlabel("X")
plt.ylabel("X^n")
plt.grid(True)
plt.show()
```

```python
# plot a scatter plot
x = [random.randint(1, 100) for _ in range(100)]
y = [random.randint(1, 100) for _ in range(100)]

plt.figure(figsize=(12,8))
plt.scatter(x, y)
plt.show()
```

```python
# plot a bar graph
plt.bar(x, y)
plt.show()
```
