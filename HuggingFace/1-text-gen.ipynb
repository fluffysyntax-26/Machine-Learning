```python
from transformers import AutoModelForCausalLM

gpt2 = AutoModelForCausalLM.from_pretrained("gpt2")
print(gpt2)
```

```python
import torch
import torch.nn.functional as F

# predict the next word
sentence = "I like machine learning to be able to do"

token_ids = tokenizer(sentence, return_tensors="pt")['input_ids']
outputs = gpt2(token_ids).logits[0, -1]
final_logits = torch.topk(outputs, 10) # sampling using topk

for index in final_logits.indices: 
    print(tokenizer.decode(index))
```

```python
sentence = "I learn machine learning to"
n = 100
while n > 0: 
    token_ids = tokenizer(sentence, return_tensors="pt")["input_ids"]
    outputs = gpt2(token_ids).logits[0, -1]
    token = tokenizer.decode(outputs.argmax())
    sentence += token
    print(n, sentence, "\n")
    n -= 1
```

```python
# different sampling strategies
def greedy_decode(logits): 
    '''Return token index with maximum probability'''
    return torch.argmax(logits, dim=-1)

tokenizer.decode(greedy_decode(outputs))

def top_k_sampling(logits, k=50): 
    '''Keeps only top-k logits, normalize them into probabilities using softmax, then sample one token from the filtered distribution'''

    values, indices = torch.topk(logits, k)
    probs = F.softmax(values, dim=-1)
    sampled = torch.multinomial(probs, 1)
    return indices[sampled]

def top_p_sampling(logits, p=0.9): 
    """ 
    Sort tokens by probability, keep smallest number who cumulative probability exceeds threshold p, then sample one token.
    """

    sorted_logits, sorted_indices = torch.sort(logits, decending=True)
    sorted_probs = F.softmax(sorted_logits, dim=-1)
    cumulative_probs = sorted_probs.cumsum(dim=-1)

    # Mask token outside nucleus
    mask = cumulative_probs > p
    sorted_logits[mask] = float("-inf")
```

```python
tokenizer.decode(top_k_sampling(outputs))
```
